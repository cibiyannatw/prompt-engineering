## Setup

Let's start by importing the necessary libraries and setting up our environment.

````python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Initialize the language model (Gemini)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")


def create_chain(prompt_template):
    """
    Create a LangChain chain with the given prompt template.

    Args:
        prompt_template (str): The prompt template string.

    Returns:
        A runnable chain (prompt | llm).
    """
    prompt = PromptTemplate.from_template(prompt_template)
    return prompt | llm
````

# Part 1: When Zero-Shot Prompting Works Well

In this section, we'll explore **successful use cases** where zero-shot prompting excels. You'll see how effective simple, well-structured prompts can be for straightforward tasks.

## 1. Direct Task Specification

In this section, we'll explore how to craft prompts that clearly define the task without providing examples. This is the essence of zero-shot prompting.

````python
direct_task_prompt = """Classify the sentiment of the following text as positive, negative, or neutral.
Do not explain your reasoning, just provide the classification.

Text: {text}

Sentiment:"""

direct_task_chain = create_chain(direct_task_prompt)

print(direct_task_chain)

# Test the direct task specification
texts = [
    "I absolutely loved the movie! The acting was superb.",
    "The weather today is quite typical for this time of year.",
    "I'm disappointed with the service I received at the restaurant."
]

for text in texts:
    result = direct_task_chain.invoke({"text": text}).content
    print(f"Text: {text}")
    print(f"Sentiment: {result}")
````

## 2. Format Specification

Providing output format guidelines in the prompt can help structure the AI's response in a zero-shot scenario.

````python
format_spec_prompt = """Generate a short news article about {topic}. 
Structure your response in the following format:

Headline: [A catchy headline for the article]

Lead: [A brief introductory paragraph summarizing the key points]

Body: [2-3 short paragraphs providing more details]

Conclusion: [A concluding sentence or call to action]"""

format_spec_chain = create_chain(format_spec_prompt)

# Test the format specification prompting
topic = "The discovery of a new earth-like exoplanet"
result = format_spec_chain.invoke({"topic": topic}).content
print(result)
````

## 3. Multi-step Reasoning

For complex tasks, we can break them down into simpler zero-shot steps. This approach can improve the overall performance of the model.

````python
multi_step_prompt = """Analyze the following text for its main argument, supporting evidence, and potential counterarguments. 
Provide your analysis in the following steps:

1. Main Argument: Identify and state the primary claim or thesis.
2. Supporting Evidence: List the key points or evidence used to support the main argument.
3. Potential Counterarguments: Suggest possible objections or alternative viewpoints to the main argument.

Text: {text}

Analysis:"""

multi_step_chain = create_chain(multi_step_prompt)

# Test the multi-step reasoning approach
text = """While electric vehicles are often touted as a solution to climate change, their environmental impact is not as straightforward as it seems. 
The production of batteries for electric cars requires significant mining operations, which can lead to habitat destruction and water pollution. 
Moreover, if the electricity used to charge these vehicles comes from fossil fuel sources, the overall carbon footprint may not be significantly reduced. 
However, as renewable energy sources become more prevalent and battery technology improves, electric vehicles could indeed play a crucial role in combating climate change."""

result = multi_step_chain.invoke({"text": text}).content
print(result)
````

## Key Takeaway: When Zero-Shot Fails

| Problem Type | Why It Fails | Fix |
|---|---|---|
| **Multi-step math** | No intermediate reasoning | Use **Chain-of-Thought** prompting |
| **Domain knowledge** | Lacks context/background | Use **Few-Shot** examples + context |
| **Strict formatting** | Adds markdown or extra text | Use **Role Prompting** + XML/explicit structure |
| **Ambiguous tasks** | Needs clarification | Use **Few-Shot** with worked examples |

**General Rule:** Zero-shot works well for **straightforward tasks** (sentiment, classification, simple generation). It fails when the model needs to **follow strict rules, learn patterns, apply specialized knowledge, or produce exact output formats**.

# Part 3: Summary & When to Use Zero-Shot

## Quick Decision Guide

Use **Zero-Shot Prompting** when:
- ✅ Task is straightforward and well-defined
- ✅ Model has sufficient knowledge about the domain
- ✅ No special reasoning patterns are required
- ✅ You need a quick solution with minimal setup

**Examples:** Sentiment analysis, text classification, simple generation, answering factual questions

---

Use **Other Techniques** when:
- ❌ Task requires multi-step reasoning → Use **Chain-of-Thought**
- ❌ Task is domain-specific → Use **Few-Shot Learning**
- ❌ Task needs specific reasoning pattern → Use **Few-Shot** + **examples**
- ❌ Task is ambiguous → Use **Role Prompting** + **constraints**

---

## Key Learning Points

| Aspect | Details |
|--------|---------|
| **Strengths** | Fast, simple, minimal setup, works for straightforward tasks |
| **Weaknesses** | No intermediate reasoning, lacks context, no pattern learning |
| **Best for** | Classification, sentiment, simple QA, generation |
| **Avoid for** | Math, logic puzzles, specialized knowledge, ambiguous tasks |

## 4. Comparative Analysis

Let's compare different zero-shot prompt structures for the same task to evaluate their effectiveness.

````python
def compare_prompts(task, prompt_templates):
    """
    Compare different prompt templates for the same task.
    
    Args:
        task (str): The task description or input.
        prompt_templates (dict): A dictionary of prompt templates with their names as keys.
    """
    print(f"Task: {task}\n")
    for name, template in prompt_templates.items():
        chain = create_chain(template)
        result = chain.invoke({"task": task}).content
        print(f"{name} Prompt Result:")
        print(result)
        print("\n" + "-"*50 + "\n")

task = "Explain concisely the concept of blockchain technology"

prompt_templates = {
    "Basic": "Explain {task}.",
    "Structured": """Explain {task} by addressing the following points:
1. Definition
2. Key features
3. Real-world applications
4. Potential impact on industries"""
}

compare_prompts(task, prompt_templates)
````

## 5. When Zero-Shot Prompting Fails

Zero-shot prompting has important limitations. It struggles with:

1. **Complex multi-step reasoning** — requires breaking problems down
2. **Domain-specific knowledge** — lacks context about specialized concepts
3. **Tasks requiring counter-intuitive thinking** — needs examples to understand patterns
4. **Ambiguous problems** — benefits from clarification examples

Let's test these failure cases:

````python
# FAILURE CASE 1: Complex Multi-Step Math Reasoning
print("=" * 70)
print("FAILURE CASE 1: Complex Multi-Step Math Problem (Zero-Shot)")
print("=" * 70)

math_problem = """A store sells apples at $3 per pound. Sarah buys 5 pounds of apples 
and gives the cashier $20. She uses the change to buy oranges at $2 each. 
How many oranges can she buy?

Answer:"""

math_chain = create_chain(math_problem)
result = math_chain.invoke({})
print(f"Zero-Shot Result:\n{result.content}")

print("\n" + "=" * 70)
print("Why It Failed:")
print("=" * 70)
print("""
The model tries to answer in one step but:
✗ Doesn't break the problem into steps
✗ Loses intermediate calculations
✗ Doesn't show the logic: apples cost = 5 × $3 = $15
✗ Doesn't calculate change: $20 - $15 = $5
✗ Doesn't determine oranges: $5 ÷ $2 = 2 oranges
""")
````

````python
# FAILURE CASE 2: Domain-Specific Knowledge Without Context
print("\n" + "=" * 70)
print("FAILURE CASE 2: Specialized Domain Knowledge (Zero-Shot)")
print("=" * 70)

bio_problem = """Explain why a mutation in the CFTR gene causes cystic fibrosis. 
What is the cellular mechanism? Use proper terminology.

Explanation:"""

bio_chain = create_chain(bio_problem)
result = bio_chain.invoke({})
print(f"Zero-Shot Result:\n{result.content}")

print("\n" + "=" * 70)
print("Why It Failed:")
print("=" * 70)
print("""
The model lacks foundational context:
✗ Doesn't explain what CFTR normally does (ion channel protein)
✗ Misses the mechanism (protein misfolding → loss of function)
✗ Skips the cellular consequence (improper ion/water balance)
✗ Oversimplifies the biological chain of causation
✗ Result: Vague or technically incorrect explanation
""")
````

````python
# FAILURE CASE 3: Strict Output Formatting Requirements
print("\n" + "=" * 70)
print("FAILURE CASE 3: Strict JSON Output (Zero-Shot)")
print("=" * 70)

strict_format_prompt = """Extract the following information and output ONLY valid JSON. 
No markdown. No explanation. Just raw JSON.

Data:
- Name: Sarah Johnson
- Age: 28
- Email: sarah.johnson@example.com
- City: San Francisco

Output:"""

strict_chain = create_chain(strict_format_prompt)
result = strict_chain.invoke({})
print(f"Zero-Shot Result:\n{result.content}")

print("\n" + "=" * 70)
print("Why It Failed:")
print("=" * 70)
print("""
Zero-shot prompts often fail with strict formatting requirements:
✗ Model wraps JSON in markdown code blocks (```json ... ```)
✗ Adds explanatory text before/after the JSON
✗ Uses camelCase instead of snake_case for keys
✗ Includes comments in the JSON output
✗ Misunderstands "ONLY JSON" instruction

Result: The output is NOT valid JSON - it's JSON wrapped in markdown!
This breaks downstream tools expecting pure JSON parsing.

Required Output:
{"name": "Sarah Johnson", "age": 28, "email": "sarah.johnson@example.com", "city": "San Francisco"}

Actual Output:
```json
{...actual output with markdown...}
```
""")
````

- Instructions are confusing or contradictory

- There's a trick or trap hidden in the problem

**Real talk:** Zero-shot is actually pretty good now. But if you're building something in production, you still need to validate the output, handle errors, and use structured constraints. Don't just trust it blindly.
